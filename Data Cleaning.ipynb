```python
import pandas as pd
```

```python
INPUT_FILE = "IE_Dataset.csv"
```

```python
OUTPUT_CSV = "IE_Dataset_Cleaned"
```

```python
OUTPUT_PARQUET = "pilot_bird_occurrences_clean.parquet"
```

```python
df=pd.read_csv(INPUT_FILE, low_memory = False)
```

```python
n_start = len(df)
```

```python
log = []
```

# Removing missing/invalid coordinates

```python
before = len(df)
```

```python
df = df[df["decimalLatitude"].notna() & df["decimalLongitude"].notna()]
log.append(f"Removed {before - len(df)} records with missing latitude/longitude")
```

```python
before =len(df)
```

```python
df = df[df["decimalLatitude"].between(-90,90) & df["decimalLongitude"].between(-180,180) & ~((df["decimalLatitude"]==0)&(df["decimalLongitude"]==0))]
```

```python
log.append(f"Removed {before - len(df)} records with out of range or 0,0 coordinates")
```

```python
df.head()
```

# Filter to Australia

```python
before = len(df)
```

```python
df = df[(df["country"] == "Australia") | (df["countryCode"] == "AU")]
```

```python
log.append(f"Removed {before - len(df)} records not flagged as Australia")
```

```python
df.head()
```

# Filter to Australia's geographic bounding box

```python
AUS_LAT_MIN, AUS_LAT_MAX = -44.0, -9.0
AUS_LON_MIN, AUS_LON_MAX = 112.0, 154.0
```

```python
before = len(df)
```

```python
df= df[df["decimalLatitude"].between(AUS_LAT_MIN, AUS_LAT_MAX) & df["decimalLongitude"].between(AUS_LON_MIN, AUS_LON_MAX)]
```

```python
log.append(f"Removed {before -len(df)} records outside Australia bounding box" f"(lat {AUS_LAT_MIN} to {AUS_LAT_MAX}, lon {AUS_LON_MIN} to {AUS_LON_MAX})")
```

```python
df.head()
```

# Coordinate Uncertainity review

```python
MAX_UNCERTAINTY_M = 10000
```

```python
before = len(df)
```

```python
df= df[ df [ "coordinateUncertaintyInMeters"].isna() | (df["coordinateUncertaintyInMeters"]<=MAX_UNCERTAINTY_M)]
```

```python
log.append(f"Removed {before - len(df)} records with coordinateUncertaintyInMeters" f">{MAX_UNCERTAINTY_M} m")
```

```python
missing_uncertainty = df["coordinateUncertaintyInMeters"].isna().sum()
```

```python
generalised = df["dataGeneralizations"].notna().sum()
```

```python
log.append(f"Note: {missing_uncertainty} remaining records have no reported" f"coordinate uncertainty (kept, flagged as unknown accuracy)")
```

```python
log.append(f"Note: {generalised} remaining records are intentionally generalised" f"(sensitive-species location masking) - kept, flagged, not treated as error")
```

```python
n_end=len(df)
```

```python
print(f"Starting records: {n_start}")
```

```python
for line in log:
    print(" -",line)
```

```python
print(f"After coordinate cleaning: {n_end} records (removed {n_start - n_end} total)")
```

# Save the intermediate WGS84 (EPSG:4326) CSV Version 

```python
df.to_csv(OUTPUT_CSV, index=False)
```

```python
print(f"Saved cleaned CSV (EPSG:4326) to {OUTPUT_CSV}")
```

# Standardise spatial reference to EPS G:3577 (GDA94/Australian Albers)

```python
pip install geopandas pyproj shapely pyarrow
```

```python
import geopandas as gpd
```

```python
gdf = gpd.GeoDataFrame( df, geometry=gpd.points_from_xy(df["decimalLongitude"],df["decimalLatitude"]),crs="EPSG:4326",)
```

```python
gdf_3577 = gdf.to_crs("EPSG:3577")
```

```python
gdf_3577["easting_m"] = gdf_3577.geometry.x
```

```python
gdf_3577["northing_m"] = gdf_3577.geometry.y
```

```python
print(f"Reprojected {len(gdf_3577)} records from EPSG:4326 to EPSG:3577")
```

# Export as Parquet

```python
gdf_3577.to_parquet(OUTPUT_PARQUET)
```

```python
print(f"Saved cleaned, projected dataset (EPSG:3577) to {OUTPUT_PARQUET}")
```
